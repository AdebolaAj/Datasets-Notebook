```python
import pandas as pd

df = pd.read_csv("sales_data.csv")

df.head()
```

```python
import matplotlib.pyplot as plt

df["Sale_Date"] = pd.to_datetime(df["Sale_Date"])

sales_over_time = (
    df.groupby("Sale_Date")["Sales_Amount"]
      .sum()
      .sort_index()
)

sales_over_time.plot(figsize=(10, 5))

plt.title("Sales Revenue Over Time")
plt.xlabel("Date")
plt.ylabel("Total Sales Amount")
plt.show()
```

```python
category_sales = (
    df.groupby("Product_Category")["Sales_Amount"]
      .sum()
      .sort_values(ascending=False)
)

category_sales.plot(kind="bar", figsize=(10, 5))

plt.title("Total Sales by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Sales Amount")
plt.xticks(rotation=45)
plt.show()
```

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
Q1 = df["Sales_Amount"].quantile(0.25)
Q3 = df["Sales_Amount"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
outliers = df[
    (df["Sales_Amount"] < lower_bound) |
    (df["Sales_Amount"] > upper_bound)
]

print("Number of outliers:", len(outliers))
outliers
```

```python
df_no_outliers = df[
    (df["Sales_Amount"] >= lower_bound) &
    (df["Sales_Amount"] <= upper_bound)
]

print("Rows before outlier removal:", len(df))
print("Rows after outlier removal:", len(df_no_outliers))

df_no_outliers.head(10)
```

```python
print("Original dataset shape:", df_no_outliers.shape)
df_no_outliers.head()
```

```python
df_sample = df_no_outliers.sample(
    frac=0.50,
    random_state=42
)

print("Original dataset shape:", df_no_outliers.shape)
print("Sampled dataset shape:", df_sample.shape)

df_sample.head()
```
